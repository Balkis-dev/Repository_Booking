```python
import pandas as pd
from pathlib import Path

class Bookcontext:

    booking_cols = [
        'Date',
        'Time',
        'Booking ID',
        'Booking Status',
        'Customer ID',
        'Vehicle Type',
        'Pickup Location',
        'Drop Location'
    ]

    performance_cols = [
        'Booking Value',
        'Ride Distance',
        'Driver Ratings',
        'Customer Rating',
        'Payment Method'
    ]

    ride_cols = [
        'Avg VTAT',
        'Avg CTAT',
        'Cancelled Rides by Customer',
        'Reason for cancelling by Customer',
        'Cancelled Rides by Driver',
        'Driver Cancellation Reason',
        'Incomplete Rides',
        'Incomplete Rides Reason'
    ]

    PATH = r"C:\Users\Dev404.DESKTOP-G7SAGDQ.000\Documents\formation\python\groupproject\Repository_Booking\ncr_ride_bookings.csv"


    def __init__(self, path:None|str=None):
        if isinstance(path, str):
            self.path = path
        else:
            self.path = Bookcontext.PATH
        return

    def csv_read(self):
        path = Path(self.path)
        return pd.read_csv(path)

    def csv_glance(self, nrows=None):
            path = Path(self.path)
            return pd.read_csv(path, nrows=nrows)
```

```python
df = Bookcontext().csv_glance(nrows=10)
display(df)
display(df.dtypes)
```
